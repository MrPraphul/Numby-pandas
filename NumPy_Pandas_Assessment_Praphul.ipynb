```python
import pandas as pd
import numpy as np




student_performance = pd.read_csv(
    "/content/drive/MyDrive/DSA/Dataset/numpy_pandas_assessment_datasets/student_performance.csv"
)



# Q11. Data Loading and Exploration

print("Q11")

print("First 5 records:")
print(student_performance.head())

print("")

print("Last 5 records:")
print(student_performance.tail())

print("")

print("Number of rows and columns:")
print(student_performance.shape)

print("")

print("Column names:")
print(student_performance.columns)

print("")

print("Data types:")
print(student_performance.dtypes)

print("")

print("Statistical summary:")
print(student_performance.describe())

print("\n")
```

```python
# Q12 (1) Final Score > 75
print("Students with Final Score greater than 75:")
display(student_performance[student_performance["Final_Score"] > 75])


# Q12 (2) Attendance < 75%
print("Students with Attendance below 75%:")
display(student_performance[student_performance["Attendance"] < 75])


# Q12 (3) Data Science AND Final Score > 80
print("Data Science students with Final Score greater than 80:")
display(
    student_performance[
        (student_performance["Department"] == "Data Science") &
        (student_performance["Final_Score"] > 80)
    ]
)


# Q12 (4) Sort by Final Score
print("Students sorted by Final Score:")
display(
    student_performance.sort_values(
        by="Final_Score",
        ascending=False
    )
)
```

```python

# Q13. Feature Creation


print("Q13")

student_performance["Average_Technical_Score"] = (
    student_performance[
        [
            "Python_Score",
            "SQL_Score",
            "Pandas_Score",
            "NumPy_Score"
        ]
    ].mean(axis=1)
)

student_performance["Performance"] = pd.cut(
    student_performance["Average_Technical_Score"],
    bins=[-np.inf, 60, 75, 90, np.inf],
    labels=[
        "Needs Improvement",
        "Average",
        "Good",
        "Excellent"
    ],
    right=False
)

print(student_performance)

print("\n")
```

```python

# Q14. GroupBy Analysis


print(" Q14 ")

department_result = student_performance.groupby(
    "Department"
).agg(
    Number_of_Students=("Student_ID", "count"),
    Average_Final_Score=("Final_Score", "mean"),
    Maximum_Final_Score=("Final_Score", "max"),
    Minimum_Final_Score=("Final_Score", "min")
)

department_result = department_result.sort_values(
    by="Average_Final_Score",
    ascending=False
)

print(department_result)

print("\n")
```

```python

# Q15. NumPy and Pandas Integration


print(" Q15 ")

scores = student_performance['Final_Score'].dropna().values

print("Mean:", scores.mean())
print("Median:", np.median(scores))
print("Std:", scores.std())
print("25th Percentile:", np.percentile(scores, 25))
print("75th Percentile:", np.percentile(scores, 75))


print(student_performance[student_performance['Final_Score'] > scores.mean()])
```

```python
#Q18
import pandas as pd
import numpy as np

df = pd.read_csv('/content/drive/MyDrive/DSA/Dataset/numpy_pandas_assessment_datasets/employee.csv')


df['Salary'] = pd.to_numeric(df['Salary'], errors='coerce')
df['Experience'] = pd.to_numeric(df['Experience'], errors='coerce')

#A. NumPy Analysis
salary = df['Salary'].dropna().values
perf = df['Performance_Score'].dropna().values

print("Mean Salary:", salary.mean())
print("Median Salary:", np.median(salary))
print("Std Salary:", salary.std())
print("Mean Performance:", perf.mean())
```

```python
#Q18
#B. Filtering
filtered = df[(df['Experience'] > 3) & (df['Performance_Score'] >= 80)]
print(filtered)
```

```python
#Q18
#C. Department Analysis
dept_summary = df.groupby('Department').agg(
    Average_Salary=('Salary', 'mean'),
    Average_Performance=('Performance_Score', 'mean'),
    Total_Projects=('Projects_Completed', 'sum'),
    Employee_Count=('Employee_ID', 'count')
)
print(dept_summary)
print("Highest Performance Dept:", dept_summary['Average_Performance'].idxmax())
```

```python
#Q18
#D. Analytical Interpretation
flagged = df[(df['Salary'] > salary.mean()) & (df['Performance_Score'] < perf.mean())]
print(flagged[['Employee_ID', 'Department', 'Salary', 'Performance_Score']])
```

```python
#Q17

df = pd.read_csv('/content/drive/MyDrive/DSA/Dataset/numpy_pandas_assessment_datasets/sales.csv')


df['Order_Date'] = pd.to_datetime(df['Order_Date'], errors='coerce')
df['Sales'] = pd.to_numeric(df['Sales'], errors='coerce')
df['Discount'] = pd.to_numeric(df['Discount'], errors='coerce')


df['Quantity'] = df['Quantity'].fillna(df['Quantity'].median())
df['Profit'] = df['Profit'].fillna(df['Profit'].median())
df['Sales'] = df['Sales'].fillna(df['Sales'].median())
df['Discount'] = df['Discount'].fillna(df['Discount'].mean())
df['Region'] = df['Region'].fillna('Unknown').str.strip().str.title()
df['Category'] = df['Category'].str.strip().str.title()
```

```python
#Q17
#B. Sales Analysis
total_sales = df['Sales'].sum()
total_profit = df['Profit'].sum()
avg_sales = df['Sales'].mean()

top_5_products = df.groupby('Product')['Sales'].sum().nlargest(5)
top_region = df.groupby('Region')['Sales'].sum().idxmax()

print(f"Total Sales: {total_sales:.2f}")
print(f"Total Profit: {total_profit:.2f}")
print(f"Average Sales: {avg_sales:.2f}")
print("Top 5 Products:\n", top_5_products)
print("Region with Highest Sales:", top_region)
```

```python
#Q17
#C. Category Analysis
cat_summary = df.groupby('Category').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Average_Discount=('Discount', 'mean')
).sort_values(by='Total_Sales', ascending=False)

print("\nCategory Summary:\n", cat_summary)
```
